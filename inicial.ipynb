### Case - Cancelamento de Clientes

Você foi contratado por uma empresa com mais de 800 mil clientes para um projeto de Dados. Recentemente a empresa percebeu que da sua base total de clientes, a maioria são clientes inativos, ou seja, que já cancelaram o serviço.

Precisando melhorar seus resultados ela quer conseguir entender os principais motivos desses cancelamentos e quais as ações mais eficientes para reduzir esse número.

Base de dados e arquivos: https://drive.google.com/drive/folders/1uDesZePdkhiraJmiyeZ-w5tfc8XsNYFZ?usp=drive_link

```python
# Passo a Passo do projeto
#pip install pandas openpyxl ipykernel nbformat plotly

# 1. Importar a base de dados
import pandas as pd

tabela = pd.read_csv("cancelamentos.csv")

# 2. Analisar a base de dados
       #entender as informações disponíveis
         #verificar se existem dados faltantes
# 2. Análise inicial (verificar colunas, tipos de dados, valores nulos, etc)
tabela = tabela.drop(columns="CustomerID")
display(tabela)
display(tabela.info())
# 3. Limpar a base de dados
         #tratar os dados faltantes
tabela = tabela.dropna()

display(tabela.info())

# 4. Analisar os dados
            #verificar a distribuição dos dados


display(tabela["cancelou"].value_counts(normalize=True))
display(tabela["duracao_contrato"].value_counts(normalize=True))


display(tabela["cancelou"].value_counts(normalize=True).map("{:.2%}".format))

#5. Análise detalhada (causa do cancelamento  dos clientes, como cada coluna impacta o cancelamento, etc)

import plotly.express as px

print(tabela.columns)


for coluna in tabela.columns:
#cria o gráfico
  grafico = px.histogram(tabela, x="duracao_contrato", color="cancelou", barmode="group", text_auto=True)

#exibe o gráfico
  grafico.show()


# 6. Conclusão

tabela_final = tabela.query("duracao_contrato != 'Monthly'")
tabela_final = tabela_final.query("ligacoes_callcenter <= 4")
tabela_final = tabela_final.query("dias_atraso <= 20")

print("-" * 30)
print(f"Taxa de cancelamento original: {tabela['cancelou'].mean():.2%}")
print(f"Taxa de cancelamento após ações: {tabela_final['cancelou'].mean():.2%}")
print("-" * 30)

#A análise de dados serviu para priorizar onde gastar dinheiro. Em vez de dar desconto para todo mundo, descobriu que o foco deve ser no Call Center e nos Contratos Mensais.
```
